## Import Librsries

```python
import pandas as pd
import numpy as np
```

## Load the Data

```python
df = pd.read_csv('Top Ranked Real Movies Dataset.csv')
df.info()
display(df.head())
```

```python
df.drop(columns=['Unnamed: 0'])
df.head()
```

## Data Analysis

```python
high_ranked_movies = df.loc[df['Movie Rating'] > 9,['Movie Name','Movie Rating']]
high_ranked_movies
```

```python
df['Year of Release'] = df['Year of Release'].str.extract(r'(\d{4})')  #4 raghsm adadi
df['Year of Release'] = pd.to_numeric(df['Year of Release'], errors = 'coerce')

df['Votes'] = df['Votes'].str.replace(',','')
df['Votes'] = pd.to_numeric(df['Votes'], errors = 'coerce')
```

```python
selected_movies = df.loc[(df['Year of Release'] > 2015) & (df['Votes'] > 1_000_00),['Movie Name','Year of Release','Votes']].head(10)
selected_movies
```

```python
Votes = df['Votes'].to_numpy()
Movie_Rating = df['Movie Rating'].to_numpy()
print(f'average of Votes: {np.mean(Votes)}')
print(f'std of votes : {np.std(Votes)}')
print(f'min of votes : {np.min(Votes)}')
print(f'max of votes : {np.max(Votes)}')

print('----------------------')
      
print(f'average of Movie rating: {np.mean(Movie_Rating)}')
print(f'std of Movie_Rating : {np.std(Movie_Rating)}')
print(f'min of Movie_Rating : {np.min(Movie_Rating)}')
print(f'max of Movie_Rating : {np.max(Movie_Rating)}')
```
