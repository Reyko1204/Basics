## Import Librsries

```python
import pandas as pd
import numpy as np
```

## Load the Data

```python
df = pd.read_csv('Top Ranked Real Movies Dataset.csv')
df.info()
display(df.head())
```

```python
df.drop(columns=['Unnamed: 0'])
df.head()
```

## Data Analysis

```python
high_ranked_movies = df.loc[df['Movie Rating'] > 9,['Movie Name','Movie Rating']]
high_ranked_movies
```

```python
df['Year of Release'] = df['Year of Release'].str.extract(r'(\d{4})')
df['Year of Release'] = pd.to_numeric(df['Year of Release'], errors = 'coerce')

df['Votes'] = df['Votes'].str.replace(',','')
df['Votes'] = pd.to_numeric(df['Votes'], errors = 'coerce')
```

```python
selected_movies = df.loc[(df['Year of Release'] > 2015) & (df['Votes'] > 1_000_00)]
selected_movies
```
