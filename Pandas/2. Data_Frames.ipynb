```python
import numpy as np
import pandas as pd
```

## Data Frame

```python
my_dict = {
    'Name': ['Fardin','Hasti','Ali','Farya'],
    'Age':[23,23,26,17],
    'City':['Kerman','Gilan','Esfahan','Ahvaz']
}
df = pd.DataFrame(my_dict)
df
```

```python
print(df.iloc[1,2])      #iloc faghat Adad
print(df.loc[1,'City'])
```

```python
print(df['Name'])
print(df[['Name','City']])     # listayr too dar too
```

```python
print(df.iloc[0])
print('\n--------\n')             
print(df.iloc[0:2])
```

```python
print(df.index)
print(df.columns)
```
