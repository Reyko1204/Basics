## import Libraries

```python
import pandas as pd
```

#### 🟦 `pd.read_csv`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |
|----------|-------------|--------------|-------------|
| `filepath_or_buffer` | مسیر یا URL فایل | بارگذاری داده | `pd.read_csv("data.csv")` |
| `sep` | جداکننده ستون‌ها | فایل‌های با `;` یا `|` | `pd.read_csv("d.csv", sep=";")` |
| `encoding` | نوع کدگذاری متن | رفع مشکل متن فارسی | `pd.read_csv("fa.csv", encoding="utf-8")` |
| `header` | ردیف عنوان ستون‌ها | فایل بدون هدر | `pd.read_csv("d.csv", header=None)` |
| `usecols` | انتخاب ستون‌ها | کاهش حافظه | `pd.read_csv("d.csv", usecols=["id","city"])` |
| `nrows` | تعداد ردیف برای خواندن | پیش‌نمایش سریع | `pd.read_csv("big.csv", nrows=1000)` |
| `chunksize` | اندازه قطعه | پردازش فایل‌های بزرگ | `for c in pd.read_csv("big.csv", chunksize=50000): pass` |


#### 🟥 `DataFrame.to_csv`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |
|----------|-------------|--------------|-------------|
| `path_or_buf` | مسیر خروجی | ذخیره CSV | `df.to_csv("out.csv")` |
| `index` | نوشتن ایندکس | حذف ایندکس در خروجی | `df.to_csv("out.csv", index=False)` |
| `encoding` | کدگذاری خروجی | حفظ متن فارسی | `df.to_csv("fa.csv", encoding="utf-8")` |

#### 🟩 `pd.read_excel`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |
|----------|-------------|--------------|-------------|
| `io` | مسیر فایل Excel | بارگذاری داده | `pd.read_excel("data.xlsx")` |
| `sheet_name` | نام یا شماره شیت | انتخاب شیت خاص | `pd.read_excel("d.xlsx", sheet_name="Sales")` |
| `usecols` | انتخاب ستون‌ها | خواندن ستون‌های ضروری | `pd.read_excel("d.xlsx", usecols="A:D")` |
| `skiprows` | پرش از ردیف‌ها | حذف توضیحات اضافی | `pd.read_excel("d.xlsx", skiprows=2)` |



#### 🟪 `DataFrame.to_excel`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |

#### 🟨 `pd.read_json`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |
|----------|-------------|--------------|-------------|
| `path_or_buf` | مسیر یا رشته JSON | بارگذاری داده | `pd.read_json("data.json")` |

| `lines` | پردازش خط‌به‌خط | فایل‌های لاگ یا NDJSON | `pd.read_json("logs.json", lines=True)` |


#### 🟫 `DataFrame.to_json`

| پارامتر | توضیح کوتاه | کاربرد اصلی | مثال کوتاه |
|----------|-------------|--------------|-------------|
| `path_or_buf` | مسیر خروجی | ذخیره JSON | `df.to_json("out.json")` |
