## Import Libraries

```python
import numpy as np 
import pandas as pd
```

## Load the Data

```python
df = pd.read_csv('Top Ranked Real Movies Dataset.csv')
df.info

df = df.drop(columns=['Unnamed: 0'])
display(df.head())
```

## Custome Functions

```python
def movie_grade(s):
    if s >= 8:
        return 'A'
    elif s <8 and s>= 6:
        return 'B'
    elif s <6 and s>= 4:
        return 'C'
    else:
        return 'D'

df['Movie Grade'] = df['Movie Rating'].apply(movie_grade)
df[['Movie Rating','Movie Grade']]
```

```python
# to get percentage
df ['Movie Rating'].apply(lambda a : np.divide(a,10)*100 )
```

```python
df['Votes'] = df['Votes'].str.replace(',','')
df['Votes'] = pd.to_numeric(df['Votes'],errors='coerce')
df['Votes'].head(3)
```

```python
df['votes_rating_ratio'] = np.divide(df['Votes'], df['Movie Rating'])
df[['Votes','Movie Rating','votes_rating_ratio']].head()
```
