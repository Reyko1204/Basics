## Import Libraries

```python
import numpy as np 
import pandas as pd
```

## Load the Data

```python
df = pd.read_csv('Top Ranked Real Movies Dataset.csv')
df.info()
display(df.head(3))
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df['Year of Release'] = df['Year of Release'].str.extract(r'(\d{4})')
df['Year of Release'] = pd.to_numeric(df['Year of Release'],errors= 'coerce')

df['Votes'] = df['Votes'].str.replace(',','')
df['Votes']= pd.to_numeric(df['Votes'],errors='coerce')
```

## Sorting Logic

```python
#df.sort_values(by='Votes', ascending= True)
# df = df.sort_values(by='Votes', ascending= False)
df.sort_values(by='Votes', ascending= False, inplace=True)
```

```python
df.sort_values(by='Votes', ascending= True, ignore_index=True)
```

```python
df.sort_values(by=['Votes','Movie Rating'],ascending=[False,True])
```
