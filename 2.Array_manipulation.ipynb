## Transpose

```python
my_arr = np.arange(1,13).reshape(3,4)
print(f'original array:\n {my_arr}\n shape: {my_arr.shape}')

my_arr_T = my_arr.T
print(f'Transposed array:\n {my_arr_T}\n shape: {my_arr_T.shape}')
```

```python
my_arr2 = np.arange(1,25).reshape(-1,3,4)
print(f'original array:\n {my_arr2}\n shape: {my_arr2.shape}')
my_arr2_T = my_arr2.transpose(2,1,0)
print(f'Transposed array:\n {my_arr2_T}\n shape: {my_arr2_T.shape}')
```

```python
new_arr = my_arr2.swapaxes(0,2)
print(f'original array:\n {new_arr}\n shape: {new_arr.shape}')
```
