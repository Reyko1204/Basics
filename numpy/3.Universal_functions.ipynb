## Universal Functions

```python
a = np.arange(1,13).reshape(3,4)
print(a)
```

```python
a * 10
```

```python
# sum
sum = a + 5
print(f'sum\n{sum}\n')

sum2 = np.add(a,5)
print(f'sum2\n{sum2}\n')

# subtract
sub = a - 10
print(f'sub\n{sub}\n')

sub2 = np.subtract(a,10)
print(f'sub2\n{sub2}\n')
```

```python
# multiple
mul = a * 10
print(f'mul\n{mul}\n')

mul2 = np.multiply(a,10)
print(f'mul2\n{mul2}\n')

# divide
div = a / 5
print(f'div\n{div}\n')

div2 = np.divide(a,5)
print(f'div2\n{div2}\n')
```

```python
# sqrt radical forje 2
sqrt = np.sqrt(a)
print(f'sqrt\n{sqrt}\n')

# log
log = np.log(a)
print(f'log\n{log}\n')

#     a
# log           'a' can't be 0 in log
#     e=2.7
```

```python
# exponential   توابع نمایی
exp = np.exp(a)
print(f'exp\n{exp}\n')

#    a
#   e                  'e' be tavan 2.7  normal sazi data



print(f'sub\n{sub}\n')          # abstract    قدر مطلق

abs = np.abs(sub)          # مقادیر منفی رو مثبت می کنه
print(f'abs\n{abs}\n')
```
