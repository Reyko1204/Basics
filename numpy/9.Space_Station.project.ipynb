## Import Libraries

```python
import numpy as np
```

1. Create and Reshape Data

```python
data = np.random.randint(60,100,40)
data
```

```python
data = data.reshape(10,4)
print(f'Raw Data:\n\n{data}\n')
```

2. Normalizing and Broadcasting

```python
var_mean = np.mean(data, axis=0)
print(f'Mean value for each health variable: {var_mean}')
data_normal = data - var_mean
print('\n Normalized Data:')
print(data_normal)
```

3. Smart Filter to Detect Danger

```python
critical_values = np.array([40,60,80,55])

mask = (data < critical_values)
print(f'people with critical Health values: {data[mask]}')
```

4. Create a Report

```python
avg_health = np.mean(data, axis=1)
print(f'mean health factors for each astronaut: \n {np.mean(data, axis=1).reshape(-1,1)}')

rankings = np.sort(avg_health)[::-1]          #az akhar be avval. az ziad be kam
print(rankings)


rankings1 = np.argsort(avg_health)[::-1] 
print(rankings1)

print('\n ranking of astronauts based on health state')
print(data[rankings1])
```
