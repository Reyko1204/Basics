## Introduction

####  **Project Story:**
A regional education director has decided to use student grade data from 4 schools to examine the overall and individual performance of students. The goals of this analysis are:
- Identify outstanding students and those who need support  
- Assess the quality of education at each school  
- Prepare a report for the educational council meeting

---

####  **Available Data:**
- 4 schools  
- 5 students per school  
- Each student has scores in 3 subjects: Math, Science, and Persian Language  
- Scores are recorded as numbers between 10 and 20

---

####  **Key Analysis Questions:**
1. What is each student's average score?  
2. Which students have a performance above 17?  
3. Which school has the highest average score?  
4. Which subject has the greatest score dispersion (standard deviation)?  
5. Do the scores need to be scaled for better comparison?

## Import Libraries

```python
import numpy as np
```

## Data Generation

```python
# Schoole -> student -> lesson
#(4, 5, 3)  3d matris
data = np.random.randint(10,21,(4,5,3))
print(f'shape of the Data: {data.shape}')
print(f'size of data : {data.size}')
print(f'Data type : {data.dtype}')

print('raw data:', '\n',data)
```

## Data preprocessing

```python
#(4,5,3)
full_students = data.reshape(-1,3)  #منفی یک خودش میفهمه چند تا ردیف باید بده و 3 تا ستون
print(f'shape of the Data: {full_students.shape}')
print(f'size of data : {full_students.size}')
print(f'Data type : {full_students.dtype}')

print('processed data:', '\n',full_students)
```

## Data Analyse

```python
data
# (4,5,3)      har 5 dar 3 yek school hesab mishe
```

```python
print('---- Dara Analyse Report -----\n')
# average score for each student

student_avg = np.mean(full_students, axis=1)
print(f'verage score for each student: \n{student_avg}\n')

print('----------------------------------\n')

# High amd Low Performance Students

high_end = np.percentile(full_students,70)
low_end = np.percentile(full_students,30)

high_performers = np.where(student_avg >= high_end)
low_performers = np.where(student_avg <= low_end)

print(f'High Performance Students: {high_performers}, grades: {student_avg[high_performers]}')
print(f'Low Performance Students: {low_performers}, grades: {student_avg[low_performers]}')

print('--------------------------------\n')

# find avg score for each schoole
schoole_avg = np.mean(data, axis=(1,2))
print(f'Average score among schooles: \n{schoole_avg}\n')
print(f'Schooles Ranking by avg_score:\n {sorted(list(schoole_avg))}')
```
