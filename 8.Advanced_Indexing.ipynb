## Indexing and Slicing

```python
a = np.random.randint(0,12,(3,4))
a
```

```python
a[0:2,1:3]
```

```python
a[ : ,[0,3]]      #ستونای وسطو نمیهوایم و سطون اخر و اول رو به دست میاریم
```

```python
a[[0,2], :]      #همون بالایی ولی اینبار برای ردیف ها
```

```python
b = np.random.randint(0,100,(4,5))
b
```

```python
rows = np.array([0,2,3])
columns = np.array([1,3,4])
print(b[rows, columns])

rows1 = [0,2,3]
columns1 = [1,3,4]
print(b[rows1,columns1])
```
