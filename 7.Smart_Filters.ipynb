## Smart Filters

```python
numbers = np.random.randint(0,13,(3,4))
numbers
```

```python
print(numbers > 5)
```

```python
numbers[numbers > 5]   # masking
```

```python
numbers[(numbers > 5) & (numbers < 8)]
```

```python
condition = (numbers > 7) | (numbers < 3)
numbers[condition]
```

```python
b = np.arange(0,100).reshape(4,25)
b
```

```python
p25 = np.percentile(b , 25)   #data e part
p75 = np.percentile(b , 75)

# ----- 25 ----- 75 -----

condition = (b >= p25) & (b <= p75)
c = b[condition]
print(c)
print (np.max(c))
print(np.min (c))
```
